An exercise: fill in the proof of disjunction commutativity.

```lean4
example (P Q : Prop) (h : P ∨ Q) : Q ∨ P := by
  rcases h with hp | hq
  · -- Here we have hp : P. Which side of Q ∨ P can use it?
    exact Or.inr hp
  · -- Here we have hq : Q. Which side can use it?
    exact Or.inl hq
```

## The same proof using `cases`

`cases` also splits `h : P ∨ Q` into its two possibilities. This version names each branch explicitly: `inl` is the P case, and `inr` is the Q case.

```lean4
example (P Q : Prop) (h : P ∨ Q) : Q ∨ P := by
  cases h with
  | inl hp => exact Or.inr hp
  | inr hq => exact Or.inl hq
```

## The same proof using `Or.elim`

`Or.elim` takes the evidence `h : P ∨ Q` and one proof for each possibility. The first function handles the P case; the second handles the Q case.

```lean4
example (P Q : Prop) (h : P ∨ Q) : Q ∨ P := by
  exact Or.elim h
    (fun hp => Or.inr hp)
    (fun hq => Or.inl hq)
```

## New terms in this chapter

- **`P`, `Q`**: Names for propositions. A proposition is a statement that can be proved.
- **`Prop`**: Lean's type for propositions.
- **`∨`**: “Or.” `P ∨ Q` means P or Q.
- **`h`**: A name for evidence supplied to a theorem or example. Here, `h : P ∨ Q` means `h` is evidence that P or Q holds.
- **`rcases`**: “Break this evidence into cases.” `rcases h with hp | hq` splits `h : P ∨ Q` into a P case and a Q case.
- **`cases`**: Splits the evidence into cases. The `inl` branch handles the left side of an `Or`; the `inr` branch handles the right side.
- **Case / branch**: One of the possibilities handled after a split. In this proof, one branch has `hp : P`; the other has `hq : Q`.
- **`hp`, `hq`**: Names chosen for the evidence in each branch: `hp : P` in the first case and `hq : Q` in the second. The names are chosen by the proof writer.
- **`exact`**: “This finishes the current goal using this proof.” Lean checks that the proof has the type the goal requires.
- **`fun`**: Starts a function. Read `fun hp => Or.inr hp` as “given `hp`, produce `Or.inr hp`.” Here it describes what to do with the proof in each case: take the evidence for `P` and use it to build the right side of `Q ∨ P`.
- **`Or`**: The name of Lean's logical “or” construction. In `Or.inr hp`, it tells us we are building evidence for a disjunction.
- **`Or.inl proof` / `Or.inr proof`**: Build evidence for the left / right side of a disjunction. For the goal `Q ∨ P`, use `Or.inr hp` when you have `hp : P`, and `Or.inl hq` when you have `hq : Q`.
- **`Or.elim`**: Handles a proof of `P ∨ Q` by giving one function for the P case and another for the Q case.
